```python
#The hashtags that were inserted was supposed to be a dollar sign. These are the prices of insurance
updated_medical_data = medical_data.replace("#", "$")
```

```python
num_records = 0

for num in updated_medical_data:
    if num == "$":
        num_records += 1
print("There are a total of " + str(num_records) + " in the data.")
```

```python
medical_data_split = updated_medical_data.split(";")

print(medical_data_split)
```

```python
medical_records = []

for data in medical_data_split:
    medical_records.append(data.split(","))
print(medical_records)
```

```python
medical_records_clean = []
# I have to do a nested loop because a regular loop cant strip the whitsapces inside the list
for record in medical_records:
  record_clean = []
  for item in record:
      record_clean.append(item.strip())
  medical_records_clean.append(record_clean)
print(medical_records_clean)
```

```python
names = []

for record in medical_records_clean:
    record[0] = record[0].upper()
    print(record[0])
    names.append(record[0])

ages = []

bmis = []

insurance_costs = []

for rec in medical_records_clean:
    ages.append(rec[1])
    bmis.append(rec[2])
    insurance_costs.append(rec[3])
#for bmi in medical_records_clean:
   # bmis.append([2])
    
#for cost in medical_records_clean:
    #insurance_cost
print("Names: " + str(names))
print("Ages: " + str(ages))
print("BMI: "  + str(bmis))
print("Insurance Costs: " + str(insurance_costs))
```

```python
total_bmi = 0.0
for whole in (bmis):
    total_bmi += float(whole)
print(total_bmi)
```

```python
average_bmi = total_bmi / len(bmis)

print(average_bmi)
```

```python
def insurance_cost_list(names, ages, bmis, insurance_costs):
    for name, age, bmi, insurance_cost in zip(names, ages, bmis, insurance_costs):
        print("{} is {} years old with a BMI of {} and an insurance cost of {}".format(name, age, bmi, insurance_cost))

    return insurance_costs

insurance_cost_list(names=names, ages=ages, bmis=bmis, insurance_costs=insurance_costs)
```
