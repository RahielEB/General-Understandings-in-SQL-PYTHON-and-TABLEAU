```python
total_cost = 0

for total in medical_cost.values():
    total_cost += total

print(total_cost)

average_cost = total_cost / len(medical_cost)

print("Average Insurance Cost: " + str(average_cost) + ".")
```

```python
names = ["Marina", "Vinay", "Connie", "Isaac", "Valentina"]
ages = [27,24,43,35,52]

zipped_ages = zip(names,ages)

name_to_ages = {names : ages for names, ages in zipped_ages}

print(name_to_ages)

```

```python
marina_age = name_to_ages.get("Marina")

print("Marina's age is "+ str(marina_age) +".")
```
