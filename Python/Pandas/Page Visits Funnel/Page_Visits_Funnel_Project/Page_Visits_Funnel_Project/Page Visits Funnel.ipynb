Page Visits Funnel:Cool T-Shirts Inc. has asked you to analyze data on visits to their website. Your job is to build a funnel, which is a description of how many people continue to the next step of a multi-step process.

In this case, our funnel is going to describe the following process:

A user visits CoolTShirts.com
A user adds a t-shirt to their cart
A user clicks “checkout”
A user actually purchases a t-shirt

```python
import pandas as pd

visits = pd.read_csv('visits.csv',
                     parse_dates=[1])
cart = pd.read_csv('cart.csv',
                   parse_dates=[1])
checkout = pd.read_csv('checkout.csv',
                       parse_dates=[1])
purchase = pd.read_csv('purchase.csv',
                       parse_dates=[1])
#Just making sure the data is acutally there
print(visits.info)
```

```python
visits_and_carts = visits.merge(cart, how = 'left')

len_of_visits_and_carts_df = len(visits_and_carts)

print(visits_and_carts.head())

print(len_of_visits_and_carts_df)
```

```python
cart_time_nulls = visits_and_carts[visits_and_carts.cart_time.isnull()]

len_cart_time_nulls = len(cart_time_nulls)
print(cart_time_nulls)
```

```python
#fomula to see how many added a t-shrit to the cart

in_cart_to_not_ratio = len_cart_time_nulls / len_of_visits_and_carts_df

print(round(in_cart_to_not_ratio,2))
```

### 83 percent of users who have visited the Cool T-Shirts.INC has not put a shirt in their cart

## Now lets do some insights to see how many put items in their cart but did not proceed to checkout

```python
# TO do this I must combine the cart and checkout DataFrames together and calculate the null values in the check out

cart_and_checkout = cart.merge(checkout, how = 'left')

print(cart_and_checkout.info())

len_cart_and_checkout = len(cart_and_checkout)
```

```python
checkout_nulls = cart_and_checkout[cart_and_checkout.checkout_time.isnull()]

len_checkout_nulls = len(checkout_nulls)

not_to_checkout_ratio = len_checkout_nulls / len_cart_and_checkout

print(round(not_to_checkout_ratio,2))
```

### From this insight we know that 35 percent did not proceed to the checkout

## Now we find the percentage of users who checked out but did not buy

```python
all_data = visits.merge(cart, how='left').merge(checkout, how='left').merge(purchase, how='left')

print(all_data.head())
```

```python
# I am making an easier way to do the same things we have done above by making a function

def calculate_the_percentages(option_1, option_2):
    percentage = len(option_1) / len(option_2)
    return percentage
```

```python
check_out_to_buy_ratio = (calculate_the_percentages(all_data[all_data.purchase_time.isnull()],all_data))

print(round(check_out_to_buy_ratio,2))
```

```python
all_data['time_to_purchase'] = \
    all_data.purchase_time - \
    all_data.visit_time
```

 ## 88% of users did not complete a purchase after reaching checkout. 

 # The checkout-to-purchase step is therefore the weakest point in the funnel. It has the highest drop-off rate of any stage. In other words, users are least likely to finish the purchase once they reach checkout.

```python
print(all_data['time_to_purchase'])

```

```python
average_time_to_purchase = all_data['time_to_purchase'].mean()

print(average_time_to_purchase)
```

### The average time it takes someone to purchase something is 43 mins
