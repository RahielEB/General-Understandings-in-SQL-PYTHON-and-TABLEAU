```python
names = ["Mohamed", "Sara", "Xia", "Paul", "Valentina", "Jide", "Aaron", "Emily", "Nikita", "Paul"]
insurance_costs = [13262.0, 4816.0, 6839.0, 5054.0, 14724.0, 5360.0, 7640.0, 6072.0, 2750.0, 12064.0]

# Adding a new person to the list(s) for insurance cost with the name and how much she pays
names.append("Priscilla")
insurance_costs.append(8320.0)

# I want to combine the files together so I save it to a variable call medical_records
medical_records_packed = zip(insurance_costs, names)
medical_records = list(medical_records_packed)

print(medical_records)
```

```python
# I want to calculate the number of medical records in the list. With a small list I could just count but with a big list I would
# have to do thisto save a lot of time.
num_medical_records = len(medical_records)

print(num_medical_records)

print("There are " + str(num_medical_records) + " medical records on file.")
```

```python
# I am checking who is the first person on the medical record I would use this in practice if I look for something specific
first_medical_record = medical_records[0]

print("Here is the first medical record: " + str(first_medical_record) + ".")
```

```python
# I am sorting the list by shortest medical cost

insurance_sort_by_least_to_most = sorted(medical_records)

print("Here are the medical records sorted by insurance cost: " + str(insurance_sort_by_least_to_most) + ".")
```

```python
# Now that the list is ordered from cheapest to expensive, I can check for the inusrances most cheapest and expensive insurances

cheapest_three = insurance_sort_by_least_to_most[:3]

print("Here arre the three cheapest insurance cost in our medical records: " + str(cheapest_three) + ".")

expensive_three = insurance_sort_by_least_to_most[-3:]

print("Here arre the three most expensive insurance cost in our medical records: " + str(expensive_three) + ".")
```
